1.1 An API, or Application Programming Interface, is a contract that defines how two programs can talk to each other. They are the backbone of how different applications communicate with each other. An API tells us what data or services are available and how to request them, but we do not need to know the internal details of how the server processes our requests. 

1.2 'REST' stands for Representational State Transfer which is a set of standard design principles used for building APIs. Following these standardized design principles allows for increased interoperability, consistency and ease of use. An API that follows the REST principles is called a RESTful API. These RESTful APIs are useful because they typically expose endpoints that map to the CRUD operations (such as Create, Retrieve, Update and Delete) which then map to HTTP methods (such as Get, Post, Put, Patch and Delete).

```python
import json 
# Create an item with a request body
item_data = {
    "name": "Laptop",
    "description": "A high-performance laptop",
    "price": 999.99,
    "quantity": 5
}

response = requests.post(
    f"{base_url}/items5/1",
    json=item_data  # Note: using json=, not params=
)
print("Created item 1:")
print(json.dumps(response.json(), indent=2))
```

```python
# Create another item
item_data2 = {
    "name": "Mouse",
    "description": "Wireless mouse",
    "price": 29.99,
    "quantity": 10
}

response = requests.post(f"{base_url}/items5/2", json=item_data2)
print("Created item 2:")
print(json.dumps(response.json(), indent=2))
```

```python
# Read an item
response = requests.get(f"{base_url}/items5/1")
print("Retrieved item 1:")
print(json.dumps(response.json(), indent=2))
```

```python
# Partial update using PATCH
update_data = {
    "price": 899.99,  # Only updating the price
    "quantity": 3     # And the quantity
}

response = requests.patch(f"{base_url}/items5/1", json=update_data)
print("Partially updated item 1:")
print(json.dumps(response.json(), indent=2))
```

```python
# List items with pagination
response = requests.get(f"{base_url}/items5?skip=0&limit=5")
print("Listed items (paginated):")
print(json.dumps(response.json(), indent=2))
```

```python
# Try to create an item with invalid data (price should be a number, not a string)
invalid_item = {
    "name": "Invalid Item",
    "price": "not a number",  # This should cause an error
    "quantity": 5
}

response = requests.post(f"{base_url}/items5/99", json=invalid_item)
print(f"Status Code: {response.status_code}")
print(f"Error Response: {response.json()}")
```
