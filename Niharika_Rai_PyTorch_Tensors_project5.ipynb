```python
mean = X.mean(dim=1)  # Tensor Mean.
sd = X.std(dim=1, unbiased=False)  # Tensor Standard Deviation.

X_np = X.numpy()  # Converted the X tensor to a numpy so that I could apply the numpy mean and standard deviation codes. 

mean_np = X_np.mean(axis=1) # NumPy Mean
sd_np = X_np.std(axis=1, ddof=0)  # NumPy Standard Deviation 

print(f"Mean: {mean}")   # Printed all the outputs 
print(f"Standard Deviation: {sd}")
print(f"Mean (numpy): {mean_np}")
print(f"Standard Deviation (numpy): {sd_np}")
```

```python
mean_ver2 = X.sum(dim=1, keepdim=True) / X.shape[1]
var = ((X - mean_ver2)**2).sum(dim=1, keepdim=True) / X.shape[1]
sd_ver2 = var ** 0.5
norm = (X - mean_ver2)/(sd_ver2 + 10**(-12))

print(f"X shape: {X.shape}")
print(f"mean_ver2 Shape: {mean_ver2.shape}")
print(f"var Shape: {var.shape}")
print(f"sd_ver2 Shape: {sd_ver2.shape}")
print(f"norm Shape: {norm.shape}")
```

```python
X_centered = X - mean_ver2
X_norm = X_centered/(sd_ver2 + 10**(-12))

print(f"X_centered Shape: {X_centered.shape}")
print(f"X_norm Shape: {X_norm.shape}")
```

3.5 From the explantions in the text, I know that broadcasting is a technique that allows users to perform element wise operations between tensors with different shapes. For example, matching the smaller tensor size to the bigger tensor size. We have this broadcasting so as to avoid any RunTimeErrors. Hence, broadcasting requires 2 rules to be followed by PyTorch. The first rule is that the dimensions must either be equal or one of them must be 1. The second rule is that if one of the dimension is 1, that is the one that gets expanded. For example, we calculated and displayed the shape of X and mean_ver2. X has a shape of (6000, 784) and mean_ver2 has a shape of (6000, 1). Since, mean_ver2 has one of its dimensions equal to 1, we know that this is one that will be expanded. Then, we also know that the 6000 in both X and mean_ver2 are equal, meeting the first rule. Hence, when we calculate 'X_centered' by subtracting mean_ver2 from X, the 1 in mean_ver2 automatically expands due to broadcasting to facilitate the calculation. Hence, when we calculate the shape of 'X_centered', we get (6000, 784) because PyTorch automatically expanded mean_ver2 from (6000, 1) to (6000, 784) to be able to subtract each row element wise. In this example, broadcasting works because comparing 784 and 1, the 1 is allowed to be expanded to 784, and comparing 60000 and 60000, they are the same size, which meets the broadcasting requirements. So, we know through this example that we can use broadcasting when computing element wise calculations. 

3.6 The second broadcasting occured when we calculated 'X_norm'. In this formula, we used X_centered which has a shape of (6000, 784) and sd_ver2 which has a shape of (6000, 1). Again, the 1 here can expand to 784 to facilitate the division element wise. The 6000 in both X_centered and sd_ver2 are equal, meeting the broadcasting requirements. Hence, when we print the shape of X_norm, we get (6000, 784) because PyTorch has automatically expanded sd_ver2 from (6000, 1) to (6000, 784) to be able to divide each row element wise.
